```python
df['Item Weight'] = df['Item Weight'].fillna(df['Item Weight'].median())
```

```python
df.columns= df.columns.str.lower()
df.columns= df.columns.str.replace(' ', '_')
```

```python
fat_content_map = {
    'LF': 'Low Fat',
    'low fat': 'Low Fat',
    'Low fat': 'Low Fat',
    'lowfat': 'Low Fat',
    'reg': 'Regular',
    'REG': 'Regular'
}
```

```python
print(df['item_fat_content'].value_counts())
```

```python
df['item_fat_content'] = df['item_fat_content'].replace(fat_content_map).str.strip().str.title()
```
